# 03 — Explainability: TreeSHAP, Adverse Action Reason Codes & Proxy Leakage Audit

**Objective:** Explain global and individual predictions using exact `TreeExplainer`, generate plain-English Adverse Action reason codes, verify explanation stability across nearest neighbors, and test whether allowed features leak `SEX` and `AGE_GROUP`.


```python
import json
import pandas as pd
from creditfair import config
from creditfair.explain import explain_applicant

with open(config.get_artifact_path('explainability_summary.json')) as f:
    exp_sum = json.load(f)

print('Top 6 Global SHAP Features:')
print(pd.DataFrame(exp_sum['global_feature_importance'][:6]).to_string(index=False))
print('\nStability Audit:', exp_sum['stability_audit'])
print('\nProxy Audit (SEX AUC):', exp_sum['proxy_audit']['SEX_proxy']['roc_auc'])
print('Proxy Audit (AGE_GROUP Macro AUC):', exp_sum['proxy_audit']['AGE_GROUP_proxy']['macro_ovr_roc_auc'])
```

### Conclusion: Proxy Leakage Proves 'Unawareness' Is Insufficient
- Allowed financial features predict `SEX` with **ROC-AUC = `0.6166`** and `AGE_GROUP` with **Macro ROC-AUC = `0.6393`**.
- Because credit limits (`LIMIT_BAL`), education, and utilization correlate strongly with age and gender, dropping protected columns does not prevent indirect demographic encoding.
